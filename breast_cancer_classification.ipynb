```python
import pandas as pd
import numpy as np

df = pd.read_csv("breast-cancer-wisconsin-data.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
print(df["diagnosis"].value_counts())
```

```python
print(df["diagnosis"].value_counts(normalize=True) * 100)
```

```python
df = df.drop(columns=["id"])
```

```python
print(df.shape)
```

```python
df["diagnosis"] = df["diagnosis"].map({"B": 0, "M": 1})
```

```python
print(df["diagnosis"].value_counts())
```

```python
# Separate features and target

X = df.drop(columns=["diagnosis"])
y = df["diagnosis"]

print("Shape of X:", X.shape)
print("Shape of y:", y.shape)
```

```python
print(X.head())
```

```python
print(X.dtypes.value_counts())
```

```python
print("Missing values in X:", X.isnull().sum().sum())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
print("Training target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
```

```python
X_test_scaled = scaler.transform(X_test)
```

```python
print("Scaled X_train:", X_train_scaled.shape)
print("Scaled X_test :", X_test_scaled.shape)
```

```python
X_train_scaled = scaler.fit_transform(X_train)
```

```python
X_test_scaled = scaler.transform(X_test)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("Scaled X_train:", X_train_scaled.shape)
print("Scaled X_test :", X_test_scaled.shape)
```

Logistic Regression -First classification model.

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)
```

```python
logistic_model.fit(X_train_scaled, y_train)
```

```python
y_pred_logistic = logistic_model.predict(X_test_scaled)
```

```python
print("First 10 predicted values:")
print(y_pred_logistic[:10])

print("\nFirst 10 actual values:")
print(y_test.values[:10])
```

```python
y_prob_logistic = logistic_model.predict_proba(X_test_scaled)[:, 1]
```

```python
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)
```

```python
log_accuracy = accuracy_score(y_test, y_pred_logistic)
log_auc = roc_auc_score(y_test, y_prob_logistic)
log_precision = precision_score(y_test, y_pred_logistic)
log_recall = recall_score(y_test, y_pred_logistic)
log_f1 = f1_score(y_test, y_pred_logistic)
log_mcc = matthews_corrcoef(y_test, y_pred_logistic)
```

```python
print("Logistic Regression Results")
print("---------------------------")
print("Accuracy :", round(log_accuracy, 4))
print("AUC      :", round(log_auc, 4))
print("Precision:", round(log_precision, 4))
print("Recall   :", round(log_recall, 4))
print("F1 Score :", round(log_f1, 4))
print("MCC      :", round(log_mcc, 4))
```

```python
from sklearn.metrics import confusion_matrix, classification_report

cm_logistic = confusion_matrix(y_test, y_pred_logistic)

print("Confusion Matrix:")
print(cm_logistic)
```

```python
print("Classification Report:")
print(
    classification_report(
        y_test,
        y_pred_logistic,
        target_names=["Benign", "Malignant"]
    )
)
```

```python
model_results = []
```

```python
model_results.append({
    "Model": "Logistic Regression",
    "Accuracy": log_accuracy,
    "AUC": log_auc,
    "Precision": log_precision,
    "Recall": log_recall,
    "F1": log_f1,
    "MCC": log_mcc
})
```

```python
import pandas as pd

results_df = pd.DataFrame(model_results)
results_df
```

Observations

Based on the actual results:

Logistic Regression: Strong overall performance with 96.49% accuracy and the highest AUC of 99.60%. It achieved high precision and recall, showing excellent ability to distinguish benign and malignant cases.

Decision Tree: Achieved 92.98% accuracy. Its performance was lower than Logistic Regression, KNN and Random Forest, with comparatively lower AUC and MCC scores.

KNN: Performed strongly with 95.61% accuracy and 97.44% precision. Scaling helped KNN because it relies on distances between observations.

Naive Bayes: Achieved 92.11% accuracy and a strong AUC of 98.91%. However, it had the lowest recall of 85.71%, indicating that it missed more malignant cases than the other models.

Random Forest: Achieved the highest accuracy of 97.37%, perfect precision of 100%, highest F1 score of 96.30%, and highest MCC of 0.9442. It produced the strongest overall performance among the evaluated models.

Random Forest: Random Forest is selected as the overall best-performing model because it achieved the highest accuracy, precision, F1 score and MCC while maintaining high recall and AUC.
